# Ensemble Blending


```python
import pandas as pd
```

```python
xgb_sub = pd.read_csv('submission.csv')
rf_sub = pd.read_csv('submission_rf.csv')
```

```python

blend = xgb_sub.copy()
blend['Heart Disease'] = (xgb_sub['Heart Disease'] + rf_sub['Heart Disease']) / 2



blend.head()
```

```python
# Convert probabilities back to binary if necessary (since the predictions might be 0.5 now)
# Any score >= 0.5 becomes 1 (Heart Disease Presence), else 0
blend['Heart Disease'] = (blend['Heart Disease'] >= 0.5).astype(int)

blend.head()
```

```python
# Save the final blended submission
blend.to_csv('submission_blend.csv', index=False)
print("Saved blended predictions to submission_blend.csv")
```
